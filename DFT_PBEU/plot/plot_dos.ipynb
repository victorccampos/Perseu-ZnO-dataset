```python
import numpy as np
import matplotlib.pyplot as plt
plt.style.use("/home/jvc/ZnO_database/mpl_themes/jvc.mplstyle")
```

```python
def read_dos(dosfile: str):
    ener, dos = np.loadtxt(dosfile, usecols=(0,1), unpack=True)
    return ener, dos
```

```python
# ---
# Total Density of States
pp_folder = "/home/jvc/ZnO_database/DFT_PBEU/04_DOS/03_ppdos"
ener, dos = read_dos(f"{pp_folder}/ZnO.dos.dat")
EFermi = 7.499 # eV
```

```python

# # ---
# # Partial Density of States
# ener1, p1_znd = read_dos(f'{pp_folder}/ZnO_PBEU.pdos_atm#1(Zn)_wfc#3(d)') # atm1
# ener2, p2_znd = read_dos(f'{pp_folder}/ZnO_PBEU.pdos_atm#2(Zn)_wfc#3(d)') # atm2
```

```python
from matplotlib.ticker import MultipleLocator, AutoMinorLocator

fig, ax = plt.subplots(figsize=(6,5), layout="constrained")
ax.set_title("GGA+U")
ax.plot(ener-EFermi, dos, c='k', label='Total', linewidth=0.5)
ax.xaxis.set_major_locator(MultipleLocator(4))

# ax.xaxis.set_minor_locator(AutoMinorLocator(2))

# Zn 3d states
# ax.plot(ener1-EFermi, p1_znd+p2_znd, c='b', label=r'$Zn_{d}$')
# ax.fill_between(ener1 - EFermi, p1_znd + p2_znd, color='b', alpha=0.3, hatch='xx')


ax.set_xlabel('Energy (eV)')
ax.set_ylabel('DOS (state/eV/unit-cell)')

ax.set_xlim([-20,20])
ax.set_ylim([0, 20])


ax.legend()
fig.savefig("Dos_pdos_d_states.pdf", bbox_inches="tight")
plt.show()
```
