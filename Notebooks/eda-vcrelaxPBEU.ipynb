````python
import pandas as pd
import matplotlib.pyplot as plt
from ase.io import read
from ase import Atoms
from pathlib import Path
````

O presente Notebook tem o intuito de analisar os dados de cálculos DFT para a determinação semi-empírica dos parâmetros de Hubbard, utilizando pseudopotenciais da biblioteca pseudo-dojo de norma conservada, na aproximação GGA-PBE.

````python
# Reading data with Atomic Simulation Environment and Pathlib.
fpaths = "/home/jvc/ZnO_database/DFT_PBEU/01_RelaxPBEU/vcrelax-files"
relax_paths: list[Path] = [f for f in Path(fpaths).iterdir() if f.suffix == ".out"]
relax_outputs: list[Atoms] = [read(fpath) for fpath in relax_paths]
````

````python
df = pd.DataFrame(
    {
        "fpath": [f for f in relax_paths],
        "fname": [f.name for f in relax_paths],
        "a": [atom.cell[0][0] for atom in relax_outputs],
        "c": [atom.cell[2][2] for atom in relax_outputs],
    }
)
print("DataFrama com dados obtidos do Atomic Simulation Environment")
display(df)
print("DataFrama com dados obtidos através de arquivo processado com bash\n[extract_bandgaps.sh => bandgaps.dat]")

df_gap = pd.read_csv("/home/jvc/ZnO_database/DFT_PBEU/01_RelaxPBEU/bandgaps.dat", sep=" ").reset_index(names="fname")
display(df_gap)

# Merge nos DataFrames pela coluna em comum "fname".
print("Merge nos DataFrames pela coluna em comum 'fname'.")
df = df.merge(right=df_gap, how="left").sort_values(by="fname").reset_index(drop=True)
display(df)
````

````python
# Filtrando pelo Band Gap Experimental e criando coluna da razão c/a para comparação experimental.
df_sorted_gaps = df.query("3.3 <= Egap <= 3.44").sort_values(by="Egap").reset_index(drop=True)
df_sorted_gaps["c/a"] = df_sorted_gaps["c"] / df_sorted_gaps["a"]

display(df_sorted_gaps)
````

Verificação de convergência do BFGS com __Bash__:


```bash
for f in *.out; do
    grep -q "bfgs converged" "$f" \
        && echo "$f : Sim" \
        || echo "$f : Não"
done
```


````python
def convergiu(fpath: Path | str) -> str:
    """Verificar a convergência do BFGS"""
    try:
        text = Path(fpath).read_text(errors="ignore")
        status = "Sim" if "bfgs converged" in text.lower() else "Não"
        return status
    except FileNotFoundError:
        "FN"
````

````python
df_sorted_gaps = df_sorted_gaps.reindex(columns=["fname","a","c","c/a", "lowest_unoccupied", "highest_occupied", "Egap","fpath"])
df_sorted_gaps
````

````python
df_sorted_gaps["Convergiu"] = df_sorted_gaps["fpath"].apply(convergiu)
display(df_sorted_gaps)
````
