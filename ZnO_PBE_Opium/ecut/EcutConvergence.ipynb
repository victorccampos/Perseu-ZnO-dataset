```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = pd.read_csv('energies.dat', names=['file', 'energy'])

data['energy'] = data['energy'].astype(float)
data['ecutwfc'] = data['file'].str.extract(r'(\d+)', expand=False).astype(int)
data['delta_E'] = np.array([0] + np.diff(data['energy']).tolist() )

data
```

```python
data.shape[0]
```

```python
for i, delta in enumerate(data['delta_E']):
    print(i)
```

```python
with plt.style.context('/home/jvc/QEspresso7.2/ZnO_database/mpl_themes/computermodern.mplstyle'):
    fig = plt.figure(figsize=(6,5), dpi=600)
    
    for i, delta in enumerate(data['delta_E']):
        if i == 0:
            plt.text(data['ecutwfc'][i]+15, data['energy'][i]-0.3, f'{delta:.2e}', fontsize=8, ha='right', rotation=0)
        elif i == 1:
            plt.text(data['ecutwfc'][i]+3, data['energy'][i]+0.3, f'{delta:.2e}', fontsize=8, ha='right', rotation=90)
        else:
            plt.text(data['ecutwfc'][i]+2, data['energy'][i]+0.3, f'{delta:.2e}', fontsize=8, ha='right', rotation=90)
    
    plt.title('NC Opium PBE-GGA')
    plt.ylabel('Energy (Ry)', labelpad=10)
    plt.xlabel('Cut-off energy (Ry)', labelpad=10)
    plt.plot(data['ecutwfc'],data['energy'], color='k')
    plt.scatter(data['ecutwfc'],data['energy'], marker='x',color='r')
    plt.xticks(range(30, 150, 10));
    plt.savefig('ecutwfc_NC-Opium.pdf')
    plt.show()
```
