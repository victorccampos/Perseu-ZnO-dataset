```python
import numpy as np
import matplotlib.pyplot as plt

# Definindo parâmetros
media = 50 # Média (centro da curva)
desvio = 5 # Desvio padrão (dispersão)
quantidade = 1000 # Número de dados gerados

# Gerando a distribuição normal
dados = np.random.normal(loc=media, scale=desvio, size=quantidade)

# Plotando o histograma
plt.hist(dados, bins=30, density=True, alpha=0.6, color='b')

# Adicionando curva de densidade (opcional)
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
p = (1 / (desvio * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - media) / desvio)**2)
plt.plot(x, p, 'k-', linewidth=2)

plt.title('Distribuição Normal')
plt.xlabel('Valores')
plt.ylabel('Densidade')
plt.show()
```
