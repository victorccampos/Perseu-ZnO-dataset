```python
import numpy as np
import pandas as pd
from pathlib import Path
```

```python
from ase import Atoms
from ase.io import read
from ase.units import Bohr
from ase.io.espresso import write_espresso_in
```

```python
df = pd.read_csv('bandgaps_filtered.dat', names=['file', 'ho','lu', 'Egap'], delimiter=" ")
df
```

```python
get_atom_obj = lambda fname : read("hubbard_scans/outputs/"+ fname)
get_a = lambda fname : read("hubbard_scans/outputs/"+ fname).cell[0][0]
get_c = lambda fname : read("hubbard_scans/outputs/"+ fname).cell[2][2]
```

```python
df['a'] = df['file'].apply(get_a)
df['c'] = df['file'].apply(get_c)
df['c/a'] = df['c'] / df['a']
```

```python
df.sort_values('a')
```

```python
ZnO_Ud12_Up7 = df['file'].loc[1]

atoms: Atoms = get_atom_obj(ZnO_Ud12_Up7)
```

```python
df.to_csv('dataframe.csv', index=False, sep=' ')
```
